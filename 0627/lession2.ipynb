延伸挑戰題
題目
讓使用者用 input() 輸入四種飲料各自的銷售杯數(原本固定的數字改成由鍵盤輸入),程式再依輸入的資料畫出圓餅圖。並自動讓「杯數最多」的那一塊突出,而不是固定突出奶茶。

提示:
input() 讀進來是文字,要怎麼轉成數字才能放進 values?
如何找出串列中最大值的位置?(可查 max() 與 index())
explode 串列要怎麼依「最大值位置」動態產生?

```python
import matplotlib.pyplot as plt

# 設定中文字型，避免中文顯示成亂碼(方框)
plt.rcParams['font.sans-serif'] = ['Microsoft JhengHei']
plt.rcParams['axes.unicode_minus'] = False

# 飲料名稱
labels = ['奶茶', '紅茶', '綠茶', '咖啡']

# 讓使用者輸入各飲料銷售杯數
sizes = []
for name in labels:
    value = int(input(f'請輸入{name}的銷售杯數：'))
    sizes.append(value)

# 顏色
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']

# 找出最大值位置，並動態產生 explode
max_index = sizes.index(max(sizes))
explode = [0.2 if i == max_index else 0 for i in range(len(sizes))]

# 建立 figure 與 axes
fig = plt.figure(figsize=(8, 8))
axes = fig.add_subplot(111)

# 畫圓餅圖
axes.pie(
    sizes,
    labels=labels,
    colors=colors,
    explode=explode,
    startangle=90,
    shadow=True,
    autopct='%1.1f%%'
)

plt.show()
```

請幫我用 Streamlit 製作一個圓餅圖（pie chart）的網頁應用程式，需求如下：

1. 使用 matplotlib 繪製圓餅圖
2. 資料內容：四個手機品牌 Nokia、Samsung、Apple、Lumia，對應數值分別為 20、30、45、10
3. 圓餅圖的顏色依序為黃色、綠色、紅色、藍色
4. 將第一塊（Nokia）使用 explode 突出顯示，突出距離 0.3
5. 加上陰影效果（shadow）
6. 每一塊要顯示百分比，格式為小數點後一位（例如 22.2%）
7. 起始角度設定為 180 度
8. 在網頁上方加一個標題文字「手機品牌市占率圓餅圖」

請給我完整的程式碼，並在最後告訴我如何用指令執行這個 Streamlit 應用程式。

```python
import streamlit as st
import matplotlib.pyplot as plt

# 頁面標題
st.title("手機品牌市占率圓餅圖")

# 中文字型設定
plt.rcParams['font.sans-serif'] = ['Microsoft JhengHei']
plt.rcParams['axes.unicode_minus'] = False

# 資料
labels = ['Nokia', 'Samsung', 'Apple', 'Lumia']
sizes = [20, 30, 45, 10]
colors = ['yellow', 'green', 'red', 'blue']
explode = [0.3, 0, 0, 0]

# 建立圖表
fig, ax = plt.subplots(figsize=(8, 8))

ax.pie(
    sizes,
    labels=labels,
    colors=colors,
    explode=explode,
    shadow=True,
    autopct='%1.1f%%',
    startangle=180
)

ax.axis('equal')

# 顯示在 Streamlit 網頁上
st.pyplot(fig)
```
