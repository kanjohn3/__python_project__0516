```python
#資料結構
#程式結構

#list
# list建立後是可變的，可以修改
score = [89,92,76,91,77]
print(score)
print(score[0])
print(score[1])
print(score[2])
print(score[3])
print(score[4])
score[0] = 100
print(score)
```

```python
#tuple
#tuple建立後是不可變的list，不可以修改，read only
#何時使用? 暫時儲存
score1 = (89,92,76,91,77)
print(score1)
print(score1[0])
print(score1[1])
print(score1[2])
print(score1[3])
print(score1[4])
#score1[0] = 100 #會報錯，tuple不可修改
score1[0] = 100
print(score1)
```

```python
#dictionary
#dictionary是key-value pair的資料結構，key是唯一的，value可以重複的
#dictionary建立後是可變的，可以修改 
student = {
    'chinese':89, 
    'english':92, 
    'math':76, 
    'science':91, 
    'history':77}
print(student)
print(student['chinese'])
print(student['english'])
print(student['math'])
print(student['science'])
print(student['history'])
```

```python
#list
scores1 = [89, 92, 76, 91, 77]
for s in scores1:
	print(s)
```

```python
#dictionary
student = {
    'chinese':89, 
    'english':92, 
    'math':76, 
    'science':91, 
    'history':77}

for key in student:
    print(student[key])    
```
